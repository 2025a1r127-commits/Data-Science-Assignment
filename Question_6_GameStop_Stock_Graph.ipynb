This notebook downloads GameStop stock data and uses `make_graph()` to plot the closing stock price with a suitable title.

```python
import yfinance as yf
import plotly.graph_objects as go
```

```python
import yfinance as yf

# Download GameStop stock data
gme_data = yf.Ticker("GME").history(period="max")
gme_data.reset_index(inplace=True)
gme_data.head()
```

```python
import plotly.graph_objects as go

def make_graph(stock_data, title):
    """Plot the closing stock price over time with the supplied graph title."""
    date_column = "Date" if "Date" in stock_data.columns else stock_data.index.name or "index"
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=stock_data[date_column] if date_column in stock_data.columns else stock_data.index,
                             y=stock_data["Close"], mode="lines", name="Close Price"))
    fig.update_layout(title=title, xaxis_title="Date", yaxis_title="Stock Price (USD)",
                      template="plotly_white")
    fig.show()
```

```python
# Plot GameStop stock data with a graph title
make_graph(gme_data, "GameStop Stock Price Over Time")
```
